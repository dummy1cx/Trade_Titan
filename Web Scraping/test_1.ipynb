```python
url = "https://jsonplaceholder.typicode.com/posts"
```

```python
try:
    response = requests.get(url)
    response.raise_for_status()

except Exception as e:
    print(e)


else:

    status_code = response.status_code
    print(f"\nstatus code: {status_code}")
    if status_code == 200:
        print("Succesful GET request!")
        posts = response.json()
        for i in range(3):
            print(f"\nPost {i + 1}: ")
            print(posts[i])

    else:
        print("Unsucceful GET request!")

          

```

```python
new_post = {
    "title" : "sample post",
    "body" : "This is testing post",
    "userId": 101
}
```

```python
try:
    response = requests.post(url, data = new_post)
    response.raise_for_status()

except Exception as e:
    print(e)

else:
    status_code = response.status_code
    if status_code == 201:
        print(f"\nStatus Code {status_code}")
        print("\nSuccessful POST request!")

        post = response.json()
        print("\nPost:")
        print(post)

    else:
        print(f"Status Code {status_code}")
        print("\nUnsuccessful POST request")
        
        
        
```

```python
#### HTML parsing
player_names = []
team_names = []
goals = []

try:
    response = requests.get(url)

    response.raise_for_status()

except Exception as e:
    print(e)

else:
    soup = BeautifulSoup(response.content, 'html.parser')
    #print("Successful!")
    players =soup.find('tbody').find_all('tr', class_="ssrcss-qqhdqi-TableRowBody e1icz100")

    for player in players:
        player_name = player.find('div', class_ ="ssrcss-m6ah29-PlayerName e1n8xy5b1").get_text(strip= True)
        team_name = player.find('div', class_ = 'ssrcss-qvpga1-TeamsSummary e1n8xy5b0').get_text(strip= True)
        goal = player.find('div', class_ = "ssrcss-18ap757-CellWrapper ef9ipf0").get_text(strip= True)
        player_names.append(player_name)
        team_names.append(team_name)
        goals.append(goal)
        #break
              
```

```python
print(player_names)
print(team_names)
print(goals)
```
