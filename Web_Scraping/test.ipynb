```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd

# setup of web driver
driver = webdriver.Chrome()
driver.maximize_window()
wait = WebDriverWait(driver, 10)

url = "https://finance.yahoo.com/"
driver.get(url)

# handle cookie popup
try:
    consent_btn = WebDriverWait(driver, 5).until(
        EC.element_to_be_clickable(
            (By.XPATH, '//*[@id="consent-page"]/div/div/div/form/div[2]/div[2]/button[1]')
        )
    )
    consent_btn.click()
    print("Cookie consent clicked successfully.")
except Exception as e:
    print("No cookie consent button found or click failed:", e)


def wait_for_page(driver, wait):
    try:
        wait.until(lambda d: d.execute_script("return document.readyState") == "complete")
        print(f'Page loaded successfully. Title: "{driver.title}"')
    except:
        print(f'Page did not fully load! Title: "{driver.title}"')


wait_for_page(driver, wait)

# Hover over the Markets menu
actions = ActionChains(driver)
try:
    market_menu = wait.until(
        EC.presence_of_element_located((
            By.XPATH,
            '/html/body/div[2]/header/div[1]/div[1]/div[1]/div[4]/div[1]/div[1]/ul[1]/li[3]/a[1]'
        ))
    )
    actions.move_to_element(market_menu).perform()
    print("Successfully hovered over 'Markets' menu.")
except Exception as e:
    print("Could not find or hover over Markets menu:", e)

# Click on Trending Tickers
try:
    trending_tickers = wait.until(
       EC.element_to_be_clickable((
           By.XPATH,
           '/html[1]/body[1]/div[2]/header[1]/div[1]/div[1]/div[1]/div[4]/div[1]/div[1]/ul[1]/li[3]/div[1]/ul[1]/li[4]/a[1]/div[1]'
       ))
    )
    driver.execute_script("arguments[0].click();", trending_tickers)
    print("Clicked on Trending Tickers.")
except Exception as e:
    print("Trending Tickers link not found or not clickable:", e)

wait_for_page(driver, wait)

# Click on Most Active
try:
    most_active = wait.until(
        EC.element_to_be_clickable((
            By.XPATH,
            '/html[1]/body[1]/div[2]/main[1]/section[1]/section[1]/section[1]/section[1]/section[1]/div[1]/div[1]/div[1]/a[1]'
        ))
    )
    driver.execute_script("arguments[0].click();", most_active)
    print("Clicked on 'Most Active' stocks.")
except Exception as e:
    print("Could not click on 'Most Active' stocks:", e)

wait_for_page(driver, wait)

data = []
page = 1

while True:
    # Scraping the main data
    wait.until(
        EC.presence_of_element_located((By.TAG_NAME, "table"))

    )

    rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")
    for row in rows:
        values = row.find_elements(By.TAG_NAME, "td")
        #print([val.text for val in values])
        stock = {
            "name": values[1].text,
            "symbol": values[0].text,
            "price": values[3].text,
            "change": values[4].text,
            "volume": values[6].text,
            "mkt_cap": values[8].text,
            "PE_ratio": values[9].text,
        }

        data.append(stock)

    # Try clicking "Next" to load more data
    try:
        next_button = wait.until(
            EC.presence_of_element_located((
                By.XPATH,
                '//*[@id="main-content-wrapper"]/section[1]/div/div[4]/div[3]/button[3]'
            ))
        )

        # Check if button exists but disabled (last page)
        if not next_button.is_enabled():
            print("Next button is disabled. Reached the last page.")
            break

        # Scroll and click via JS
        driver.execute_script("arguments[0].scrollIntoView(true);", next_button)
        driver.execute_script("arguments[0].click();", next_button)
        print(f"Clicked Next. Moving to page {page + 1}.")
        page += 1
        time.sleep(2)

    except Exception as e:
        print("No next button found or not clickable.")
        print(f"Reason: {e.__class__.__name__}")
        break

# Convert to DataFrame and save
df = pd.DataFrame(data)
print(f"Scraped {len(df)} rows total.")
df.to_csv("yahoo_most_active_stocks.csv", index=False)
print("Data saved to 'yahoo_most_active_stocks.csv'.")

driver.quit()
```
